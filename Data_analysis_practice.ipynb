```python
#Data cleaning
#.str.replace- replaces values within  a sting and .astype converts the data type to a float
dd['Sales'] = dd['Sales'].str.replace('$','',regex= False).str.replace(',','',regex= False).astype(float)
```

```python
#call a column
dd['Discounts']
```

```python
#for gross sales, manufacturing price, COGS, sale price, discount
#.str.strip removes leading and trailing white spaces
#replaces empty string with null values then convert the column to a float
#dd['Gross Sales'] = dd['Gross Sales'].str.replace('$','').str.replace(',','').astype(float)
#dd['Manufacturing Price'] = dd['Manufacturing Price'].str.replace('$','').str.replace(',','').astype(float)
#dd['COGS'] = dd['COGS'].str.replace('$','').str.replace(',','').astype(float)
#dd['Sale Price'] = dd['Sale Price'].str.replace('$','').str.replace(',','').astype(float)
dd['Discounts'] = dd['Discounts'].str.strip().str.replace('$','').str.replace(',','').str.replace('-','')
dd['Discounts'] = dd['Discounts'].replace('',np.nan).astype(float)
```

```python
#replace the null values with the mean value in the discount column
dd['Discounts'] = dd['Discounts'].fillna(dd['Discounts'].mean())
dd.head()
```

```python
#data entries in the country column are actually states so we need to rename the column
dd.rename(columns= {'Country': 'State'}, inplace=True)
```

```python
#create a new column named month then let the extracted month form the date be the entries in this column
dd['Month']= dd['Date'].dt.month_name()
display(dd)
```

```python
dd['Day']= dd['Date'].dt.day_name()
```

```python
import matplotlib.pyplot as plt
dd.groupby(['Segment']).agg({'Sales':'sum'}).plot(kind='bar')
plt.title('Sale Perfomance by Segement')
plt.xlabel('Segment')
plt.ylabel('Sum of Sales')
dd.groupby(['Product']).agg({'Sales':'sum'}).plot(kind='bar')
plt.title('Sale Perfomance by Product')
plt.xlabel('Product')
plt.ylabel('Sum of Sales')
```
